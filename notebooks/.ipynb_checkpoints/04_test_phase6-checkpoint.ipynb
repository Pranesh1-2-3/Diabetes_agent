# Memory-Enabled Phase 6 Testing

This notebook extends Phase 6 testing with memory caching:
1. Before running analysis, check memory for similar cases
2. If similar case found with high confidence, return cached result
3. If no match, run full analysis and store in memory
4. Store synopsis and outcome for future reference

```python
# Initialize memory system safely
import sys, os
from pathlib import Path
sys.path.append(os.path.abspath(os.path.join(os.getcwd(), "..")))
from diabetes_services.memory.utils import CaseMemoryManager
from sentence_transformers import SentenceTransformer

# Reinitialize memory system
memory_manager = CaseMemoryManager()  # Creates new index + metadata
encoder = SentenceTransformer('all-MiniLM-L6-v2')

print(f"Memory system initialized with {memory_manager.index.ntotal} stored cases")

```

```python
import os
import json
import groq

async def analyze_with_groq(patient_data, classifier_output, retrieved_chunks):
    """Analyze patient data, model output, and retrieved knowledge using Groq LLM."""
    if "GROQ_API_KEY" not in os.environ:
        api_key = input("Enter your Groq API key: ")
        os.environ["GROQ_API_KEY"] = api_key

    groq_client = groq.Groq(api_key=os.environ["GROQ_API_KEY"])

    # Build context
    context = f"""
    Patient Data:
    {json.dumps(patient_data, indent=2)}

    Classifier Output:
    {json.dumps(classifier_output, indent=2)}

    Retrieved Medical Guidelines:
    """

    for i, chunk in enumerate(retrieved_chunks, 1):
        context += f"\nChunk {i}:\n"
        context += f"Document: {chunk['doc_id']}, Page: {chunk['page_num']}\n"
        context += f"Content: {chunk['text']}\n"

    system_prompt = """You are a medical analysis assistant. Analyze the patient data, classifier output, and medical guidelines provided.
    Your response must be in JSON format with the following structure:
    {
        "conclusion": "Clear summary of patient's condition and risk factors",
        "reasoning": ["Step 1 of analysis", "Step 2", "etc"],
        "next_steps": ["Recommended action 1", "Action 2", "etc"],
        "sources": [
            {
                "doc_id": "Source document ID",
                "page": page_number,
                "quote": "Exact quote from source",
                "relevance": "Why this source supports the analysis"
            }
        ]
    }

    Always cite specific guidelines and include exact quotes when making recommendations."""

    # Call Groq API
    chat_completion = groq_client.chat.completions.create(
        messages=[
            {"role": "system", "content": system_prompt},
            {"role": "user", "content": context}
        ],
        model="llama-3.1-8b-instant",
        temperature=0.3,
        max_tokens=2048
    )

    # Parse the JSON response safely
    try:
        result = json.loads(chat_completion.choices[0].message.content)
    except Exception as e:
        print("Groq output parsing failed:", e)
        print("Raw content:", chat_completion.choices[0].message.content)
        return None

    return result
```

```python
async def analyze_case_with_memory(patient_data: dict, min_similarity: float = 0.99):
    """Run analysis with memory caching, enforcing similarity threshold."""
    import numpy as np, faiss, json

    # 1️⃣ Generate case summary
    case_summary = f"""
    Patient data: BMI {patient_data.get('BMI')}, 
    blood pressure {patient_data.get('BloodPressure')}, 
    glucose {patient_data.get('Glucose')},
    age {patient_data.get('Age')},
    diabetes pedigree {patient_data.get('DiabetesPedigreeFunction')}
    """

    # 2️⃣ Generate embedding
    embedding = encoder.encode([case_summary])[0]                     # 1D array
    embedding_np = np.array(embedding, dtype=np.float32).reshape(1, -1)  # 2D for FAISS query

    # 3️⃣ Check FAISS index dimension
    expected_dim = embedding_np.shape[1]
    index_dim = getattr(memory_manager.index, "d", None)
    if index_dim != expected_dim:
        print(f"[⚙️ Rebuilding FAISS index] Mismatch detected: index dim={index_dim}, encoder dim={expected_dim}")
        memory_manager.index = faiss.IndexFlatL2(expected_dim)
        if hasattr(memory_manager, "cases"):
            memory_manager.cases = []
        print("[✅] FAISS index rebuilt with correct dimension")

    # 4️⃣ Query memory for similar cases
    similar_cases = memory_manager.query_similar_cases(
        embedding=embedding_np,
        k=2,
        min_similarity=min_similarity
    )
    classifier_output = predict_risk(patient_data)


    # 5️⃣ Return cached result if threshold met
    if similar_cases:
        cached_case = similar_cases[0]
        if cached_case.get('synopsis') and cached_case['outcome'] == classifier_output['prediction']:  # complete analysis
            print(f"Found cached analysis (similarity: {cached_case['similarity']:.3f})")
            AuditLogger.log_request(
                request_type="case_analysis",
                patient_data=memory_manager.anonymize_case(patient_data),
                model_versions=cached_case['data'].get('model_versions', {}),
                memory_id=cached_case['case_id'],
                cached_result=True
            )
            return {
                'cached': True,
                'case_id': cached_case['case_id'],
                'similarity': cached_case['similarity'],
                'synopsis': cached_case['synopsis'],
                'outcome': cached_case['outcome']
            }

    # 6️⃣ No cached hit → run full pipeline
    print("No cached analysis found, running full pipeline...")

    model_versions = {
        "classifier": "xgboost-v1.0",
        "embedding": "all-MiniLM-L6-v2",
        "llm": "mixtral-8x7b-32768"
    }

    # Get classifier prediction
    classifier_output = predict_risk(patient_data)
    print("\nClassifier output:", json.dumps(classifier_output, indent=2))

    # Get relevant guidelines
    query = build_search_query(patient_data, classifier_output)
    retrieved_chunks = await search_guidelines(query)
    print(f"\nRetrieved {len(retrieved_chunks)} relevant chunks")

    # Run LLM analysis
    groq_analysis = await analyze_with_groq(
        patient_data=patient_data,
        classifier_output=classifier_output,
        retrieved_chunks=retrieved_chunks
    )

    # 7️⃣ Create synopsis
    synopsis = {
        "risk_level": classifier_output["prediction"],
        "confidence": classifier_output["probability"],
        "conclusion": groq_analysis["conclusion"],
        "next_steps": groq_analysis["next_steps"],
        "evidence": [
            {"source": src["doc_id"], "quote": src["quote"]}
            for src in groq_analysis["sources"][:2]  # top 2 sources
        ]
    }

    # 8️⃣ Cache the new case
    case_id = memory_manager.upsert_case(
        case_data=patient_data,
        embedding=embedding.astype(np.float32),  # store 1D float32
        model_versions=model_versions,
        outcome=classifier_output["prediction"],
        synopsis=synopsis
    )

    # Log full analysis
    AuditLogger.log_request(
        request_type="case_analysis",
        patient_data=memory_manager.anonymize_case(patient_data),
        model_versions=model_versions,
        classifier_output=classifier_output,
        retrieved_docs=retrieved_chunks,
        memory_id=case_id,
        cached_result=False
    )

    return {
        'cached': False,
        'case_id': case_id,
        'classifier_output': classifier_output,
        'retrieved_chunks': retrieved_chunks,
        'analysis': groq_analysis,
        'synopsis': synopsis
    }
```

```python
import numpy as np

def make_json_serializable(obj):
    """Recursively convert numpy and other non-serializable types."""
    if isinstance(obj, dict):
        return {k: make_json_serializable(v) for k, v in obj.items()}
    elif isinstance(obj, list):
        return [make_json_serializable(v) for v in obj]
    elif isinstance(obj, (np.float32, np.float64)):
        return float(obj)
    elif isinstance(obj, (np.int32, np.int64)):
        return int(obj)
    else:
        return obj
```

```python
def build_search_query(patient_data: dict, classifier_output: dict) -> str:
    """Build a search query based on patient data and classifier output."""
    risk_level = classifier_output["prediction"]
    
    query_parts = [
        f"guidelines for {risk_level} diabetes patient",
        f"BMI {patient_data.get('BMI', 'N/A')}",
        f"blood pressure {patient_data.get('BloodPressure', 'N/A')}",
        f"glucose {patient_data.get('Glucose', 'N/A')}"
    ]
    
    return " ".join(query_parts)

# Get relevant chunks
async def get_relevant_chunks(query: str):
    """Get relevant chunks from RAG API."""
    try:
        chunks = await search_guidelines(query)
        return chunks
    except Exception as e:
        print(f"Error retrieving chunks: {e}")
        return []
```
